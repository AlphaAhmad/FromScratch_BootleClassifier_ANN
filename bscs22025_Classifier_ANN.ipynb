## Dividing into Train, Test Split

The Issue we faced here that the shutil.copy() function requires that both the source and destination paths given are of files but 

```python
import os
import shutil
import random

def split_dataset(input_dir, output_dir, train_split=0.7, val_split=0.15, test_split=0.15):
    # Check splits sum to 1
    assert train_split + val_split + test_split == 1, "Splits must sum to 1"

    # Ensure required folders exist
    required_folders = ['train', 'val', 'test']
    for folder in required_folders:
        folder_path = os.path.join(output_dir, folder)
        if not os.path.exists(folder_path):
            raise FileNotFoundError(f"Required folder '{folder}' does not exist in {output_dir}. Please create it manually.")

    # Process images
    all_images = []  # List of all images with their class labels

    for class_name in os.listdir(input_dir):
        class_dir = os.path.join(input_dir, class_name)
        if not os.path.isdir(class_dir):
            continue  # Skip non-directory files
        images = [img for img in os.listdir(class_dir) if os.path.isfile(os.path.join(class_dir, img))]  # Only files
        random.shuffle(images)

        # Add images and their class labels to the list
        all_images.extend([(os.path.join(class_dir, img), class_name) for img in images])

    # Shuffle all images
    random.shuffle(all_images)

    # Calculate split sizes
    total_images = len(all_images)
    train_size = int(train_split * total_images)
    val_size = int(val_split * total_images)

    train_images = all_images[:train_size]
    val_images = all_images[train_size:train_size + val_size]
    test_images = all_images[train_size + val_size:]

    # Copy images to corresponding directories
    for idx, (img_path, class_name) in enumerate(train_images):
        print(f"Copying {img_path} to train folder...")
        shutil.copy(img_path, os.path.join(output_dir, 'train', f"{class_name}_{idx}.jpg"))

    for idx, (img_path, class_name) in enumerate(val_images):
        print(f"Copying {img_path} to val folder...")
        shutil.copy(img_path, os.path.join(output_dir, 'val', f"{class_name}_{idx}.jpg"))

    for idx, (img_path, class_name) in enumerate(test_images):
        print(f"Copying {img_path} to test folder...")
        shutil.copy(img_path, os.path.join(output_dir, 'test', f"{class_name}_{idx}.jpg"))

    print("Dataset split completed successfully.")

# Paths
input_directory = r"C:\Users\Muhammad\Desktop\semester 5\AI\Bscs22025_Project-dataset-modified\Data"
output_directory = r"C:\Users\Muhammad\Desktop\semester 5\AI\Bscs22025_Project-dataset-modified\Data\split"

# Split the dataset
split_dataset(input_directory, output_directory)
```

## Step 2: Now Extracting Features
### Using the library CNN to do this since it is permitted

1. 1st we will preprocess the image so that we can change it to 128x128 and also keep its features preseveed since the CNN takes a square image

```python
import os
import cv2
import numpy as np
from tensorflow.keras.applications import MobileNetV2

def preprocess_image(image_path, target_size=(128, 128)):
    """
    Preprocesses an image:
    1. Resize while maintaining the aspect ratio
    2. Pad or crop the image to the target size
    3. Normalize pixel values
    
    Parameters:
        image_path (str): Path to the input image.
        target_size (tuple): Desired output size (height, width).
    
    Returns:
        np.ndarray: Preprocessed image.
    """
    img = cv2.imread(image_path)
    if img is None:
        raise ValueError(f"Image not found at path: {image_path}")
    
    original_height, original_width = img.shape[:2]
    aspect_ratio = original_width / original_height

    # Resize while maintaining the aspect ratio
    if aspect_ratio > 1:  # Wider than tall
        new_width = target_size[1]
        new_height = int(new_width / aspect_ratio)
    else:  # Taller than wide
        new_height = target_size[0]
        new_width = int(new_height * aspect_ratio)

    resized_img = cv2.resize(img, (new_width, new_height))

    # Pad or crop to achieve target size
    padded_img = cv2.copyMakeBorder(
        resized_img,
        (target_size[0] - new_height) // 2,
        (target_size[0] - new_height + 1) // 2,
        (target_size[1] - new_width) // 2,
        (target_size[1] - new_width + 1) // 2,
        cv2.BORDER_CONSTANT,
        value=(0, 0, 0),  # Black padding
    )

    padded_img = cv2.cvtColor(padded_img, cv2.COLOR_BGR2RGB)
    return padded_img / 255.0  # Normalize pixel values to [0, 1]

def extract_features(image, cnn_model):
    # Preprocessed image of 128 x 128 into the CNN for feature extraction
    feature_map = cnn_model.predict(np.expand_dims(image, axis=0))
    return feature_map.flatten() # reducing dimension into 1 so we can feed it to Neural Network

```
